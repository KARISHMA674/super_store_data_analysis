```python
import pandas as pd
import plotly.io as pio
import plotly.express as px
import plotly.graph_objects as go
import plotly.colors as colors

pio.templates.default="plotly_white"
data=pd.read_csv("/content/superstore.csv")
print(data)
```

```python
print(data.head())
```

```python
print(data.describe())
```

```python
print(data.info())
```

```python
print(data.isnull().sum())
```

```python
#converting data columns.
data["Order Date"]=pd.to_datetime(data["Order Date"])
print(data)
```

```python
data["Ship Date"]=pd.to_datetime(data["Ship Date"])
print(data)
```

```python
data["Order Month"]=data["Order Date"].dt.month
print(data)
```

```python
data["Order Year"]=data["Order Date"].dt.year
print(data)
```

```python
data["Order Day of Week"]=data["Order Date"].dt.day_of_week
print(data)
```

```python
sales_by_month=data.groupby('Order Month')['Sales'].sum().reset_index()
print(sales_by_month)
```

```python
fig=px.line(sales_by_month,x='Order Month',y='Sales',title='Monthly Sales Analysis')
fig.show()
```

```python
#sales by category
sales_by_category=data.groupby('Category')['Sales'].sum().reset_index()
print(sales_by_category)
```

```python
fig=px.pie(sales_by_category,values='Sales',names='Category',hole=0.2,color_discrete_sequence=px.colors.qualitative.Pastel)
fig.show()
```

```python
#profit analysis by subcategory
profit_by_subcategory=data.groupby('Sub-Category')['Profit'].sum().reset_index()
fig=px.bar(profit_by_subcategory,x='Sub-Category',y='Profit',title="Profit Analysis by Sub-Category")
fig.show()
```
